## 1. head()

The `head()` function displays the first few rows of a DataFrame.
By default, it displays the first 5 rows and helps us quickly inspect the dataset.

### Example

In an AI/ML project, after loading a dataset, we can use `head()`
to quickly check what the first few records look like.

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Tony", "John", "David", "Sam", "Alex", "Ravi"],
    "Age": [24, 25, 23, 26, 22, 27],
    "Score": [85, 90, 78, 88, 92, 80]
})

df.head()
```

### Output Explanation

The first 5 rows of the DataFrame are displayed.
We can also use `df.head(3)` to display the first 3 rows.

## 2. tail()

The `tail()` function displays the last few rows of a DataFrame.
By default, it displays the last 5 rows.

### Example

After processing a dataset, `tail()` can be used to check
the last records and verify that the data was loaded correctly.

```python
df.tail()
```

### Output Explanation

The last 5 rows of the DataFrame are displayed.
We can use `df.tail(3)` to display the last 3 rows.

## 3. sample()

The `sample()` function randomly selects rows from a DataFrame.
It is useful for checking random records instead of only the beginning or end.

### Example

When working with a large dataset, we can randomly inspect a few records
to understand whether the data contains different types of values.

```python
df.sample(3, random_state=42)
```

### Output Explanation

Three random rows are displayed.
`random_state=42` makes the random selection reproducible.

## 4. info()

The `info()` function provides a summary of a DataFrame.
It shows the number of rows, columns, data types, and non-null values.

### Example

Before preprocessing an ML dataset, `info()` helps identify
missing values and incorrect data types.

```python
df.info()
```

### Output Explanation

The output shows the DataFrame structure, number of entries,
column names, non-null counts, and data types.

## 5. describe()

The `describe()` function generates descriptive statistics for numerical columns.
It provides count, mean, standard deviation, minimum, maximum, and percentiles.

### Example

In a sales dataset, `describe()` can help understand the distribution
of numerical features such as price, quantity, and sales.

```python
df.describe()
```

### Output Explanation

The output provides summary statistics for numerical columns.
These statistics help us understand the general behavior of the data.

## 6. shape

The `shape` attribute returns the number of rows and columns in a DataFrame.
The result is represented as `(rows, columns)`.

### Example

Before training an ML model, we can check the dataset size
to understand how many records and features are available.

```python
df.shape
```

### Output Explanation

The output shows the number of rows and columns.
For example, `(6, 3)` means 6 rows and 3 columns.

## 7. columns

The `columns` attribute returns the names of all columns in a DataFrame.
It helps us understand the features available in the dataset.

### Example

In an ML dataset, we can use `columns` to identify
which features are available before selecting them.

```python
df.columns
```

### Output Explanation

The output displays all column names in the DataFrame.

## 8. dtypes

The `dtypes` attribute shows the data type of each column.
It helps us understand how Pandas represents each feature.

### Example

Before preprocessing, we can check whether columns contain
integers, floating-point values, strings, or other data types.

```python
df.dtypes
```

### Output Explanation

The output shows the data type of each column.
Checking data types helps prevent errors during data processing.

## 9. memory_usage()

The `memory_usage()` function shows how much memory each column uses.
It is useful when working with large datasets.

### Example

An AI/ML dataset may contain millions of rows.
Checking memory usage helps us identify columns that consume too much memory.

```python
df.memory_usage(deep=True)
```

### Output Explanation

The output shows the memory used by each column in bytes.
`deep=True` provides a more accurate memory calculation for object/string data.

## 10. nunique()

The `nunique()` function returns the number of unique values in each column.
It is useful for understanding the variety of values in categorical features.

### Example

In customer data, we may want to know how many unique ages,
cities, or product categories are present.

```python
df.nunique()
```

### Output Explanation

The output shows the number of unique values in each column.
This can help identify categorical features and high-cardinality columns.

## 11. value_counts()

The `value_counts()` function counts how frequently each unique value occurs.
It is mainly used for analyzing categorical columns.

### Example

A company can use `value_counts()` to find how many customers
belong to each city or product category.

```python
customer_data = pd.DataFrame({
    "City": ["Chennai", "Madurai", "Chennai", "Trichy", "Chennai", "Madurai"]
})

customer_data["City"].value_counts()
```

### Output Explanation

The output shows the frequency of each city.
For example, Chennai appears more frequently than the other cities.
