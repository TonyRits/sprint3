## 1. What is Pandas?

Pandas is a Python library used for data manipulation and data analysis.
It provides useful data structures such as Series and DataFrame for working with structured data.

### Example

A company has customer data containing names, ages, and purchase amounts.
Pandas can be used to load, analyze, filter, and transform this data easily.

```python
import pandas as pd

data = {
    "Name": ["Tony", "John", "David"],
    "Age": [24, 25, 23],
    "Purchase": [500, 700, 450]
}

df = pd.DataFrame(data)

df
```

### Output Explanation

The DataFrame displays customer information in rows and columns.
Pandas makes structured data easier to analyze and manipulate.

## 2. Why Pandas?

Pandas is used because it provides simple and powerful tools for handling structured data.
It supports filtering, grouping, cleaning, analysis, and file operations.

### Example

In an AI/ML project, Pandas can be used to load a dataset,
check missing values, select features, and prepare data before model training.

```python
import pandas as pd

sales = pd.DataFrame({
    "Product": ["Laptop", "Phone", "Tablet"],
    "Sales": [50000, 30000, 20000]
})

print(sales)
print("\nAverage Sales:", sales["Sales"].mean())
```

### Output Explanation

Pandas displays the sales data and calculates the average sales.
This shows how Pandas can be used for basic data analysis.

## 3. Series

A Series is a one-dimensional labeled data structure in Pandas.
It can store values such as numbers, strings, or other Python objects.

### Example

A company can store the monthly sales values of a product in a Pandas Series.

```python
import pandas as pd

sales = pd.Series([10000, 15000, 12000, 18000])

print(sales)
```

### Output Explanation

The output contains sales values with an index starting from 0.
A Series represents a single column or one-dimensional data.

## 4. DataFrame

A DataFrame is a two-dimensional labeled data structure in Pandas.
It contains rows and columns and is commonly used to represent datasets.

### Example

A customer dataset can contain multiple columns such as Name, Age, City, and Purchase.
This type of structured data can be represented using a DataFrame.

```python
import pandas as pd

data = {
    "Name": ["Tony", "John", "David"],
    "Age": [24, 25, 23],
    "City": ["Chennai", "Madurai", "Trichy"]
}

df = pd.DataFrame(data)

df
```

### Output Explanation

The DataFrame displays multiple columns and rows.
Each row represents a record and each column represents a feature or attribute.

## 5. Creating DataFrames

A DataFrame can be created from dictionaries, lists, NumPy arrays, or existing datasets.
The most common method is creating it from a dictionary.

### Example

Employee information such as employee name, department, and salary
can be converted into a DataFrame for analysis.

```python
import pandas as pd

data = {
    "Employee": ["Tony", "John", "David"],
    "Department": ["AI", "HR", "Sales"],
    "Salary": [40000, 35000, 30000]
}

df = pd.DataFrame(data)

df
```

### Output Explanation

The dictionary is converted into a structured DataFrame.
Each dictionary key becomes a column.

## 6. Data Types

Pandas assigns data types to columns based on the values they contain.
Common types include integer, float, object/string, boolean, and datetime.

### Example

In a customer dataset, Age may be an integer,
Salary may be a float, and Customer_Name may be a string.

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Tony", "John", "David"],
    "Age": [24, 25, 23],
    "Salary": [40000.5, 35000.0, 30000.5]
})

print(df.dtypes)
```

### Output Explanation

The output shows the data type of each column.
Understanding data types is important before performing data analysis or ML preprocessing.

## 7. Index

An index is used to identify and access rows in a Pandas DataFrame or Series.
By default, Pandas assigns numerical indexes starting from 0.

### Example

A customer dataset can use customer IDs as the index instead of default numbers.

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Tony", "John", "David"],
    "Purchase": [500, 700, 450]
})

df.index = ["C101", "C102", "C103"]

df
```

### Output Explanation

The default numerical index is replaced with customer IDs.
The index helps identify individual rows.

## 8. Columns

Columns represent individual features or variables in a DataFrame.
Each column contains values of a particular attribute.

### Example

In a customer dataset, Name, Age, and Purchase can be separate columns.

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Tony", "John", "David"],
    "Age": [24, 25, 23],
    "Purchase": [500, 700, 450]
})

print(df.columns)
```

### Output Explanation

The output displays all column names in the DataFrame.
Columns are commonly used as features during data analysis and machine learning.

## 9. Reading Data

Pandas can read data from different file formats such as CSV, Excel, JSON, and SQL databases.
The read functions convert external data into a DataFrame.

### Example

A sales CSV file can be loaded into Pandas before performing data analysis.

```python
import pandas as pd

df = pd.read_csv("sales.csv")

print(df.head())
```

### Output Explanation

The CSV file is loaded into a Pandas DataFrame.
The `head()` function displays the first five rows of the dataset.

## 10. Writing Data

Pandas can save DataFrames into different file formats such as CSV, Excel, and JSON.
This is useful for storing processed or analyzed data.

### Example

After cleaning a dataset, we can save the processed data as a CSV file
and use it later for machine learning or reporting.

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Tony", "John", "David"],
    "Score": [85, 90, 78]
})

df.to_csv("processed_data.csv", index=False)

print("Data saved successfully.")
```

### Output Explanation

The DataFrame is saved as `processed_data.csv`.
The `index=False` prevents Pandas from saving the DataFrame index as an additional column.
