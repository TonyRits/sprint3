# 1. Datetime

Datetime represents date and time values in Python and Pandas.
Pandas provides `pd.to_datetime()` to convert values into datetime format.

## Example

A date such as "2026-10-08" can be converted into Pandas datetime format.

```python
import pandas as pd

date = pd.to_datetime("2026-10-08")

print(date)
print(type(date))
```

## Output Explanation

The date is converted into Pandas datetime format.
This allows us to perform date-based operations easily.

#  2. Timestamp

A Timestamp represents a specific date and time.
Pandas provides `pd.Timestamp()` to create timestamp values.

## Example

We can create a timestamp containing both date and time.

```python
import pandas as pd

timestamp = pd.Timestamp("2026-10-08 10:30:00")

print(timestamp)
print(timestamp.year)
print(timestamp.month)
print(timestamp.day)
```

## Output Explanation

The output shows the complete date and time.
We can also extract the year, month, and day from the timestamp.

# 3. Date Parsing

Date parsing means converting date values from strings into datetime format.
It is important when working with real-world datasets.

## Example

A dataset may contain dates as text such as "08-10-2026".
We can convert these text values into datetime values.

```python
import pandas as pd

dates = ["08-10-2026", "09-10-2026", "10-10-2026"]

df = pd.DataFrame({"Date": dates})

df["Date"] = pd.to_datetime(df["Date"], dayfirst=True)

print(df)
print(df.dtypes)
```

## Output Explanation

The Date column is converted from string type to datetime64 type.
Now we can perform date calculations and extract date features.

# 4. Timedelta

Timedelta represents the difference between two dates or times.
It can be used to calculate durations such as days, hours, and minutes.

## Example

We can calculate the number of days between two dates.

```python
import pandas as pd

start_date = pd.Timestamp("2026-10-01")
end_date = pd.Timestamp("2026-10-08")

difference = end_date - start_date

print(difference)
print(difference.days)
```

## Output Explanation

The difference between the two dates is 7 days.
Timedelta is useful for calculating time intervals.

# 5. Time Zones

Time zones represent different geographical time standards.
Pandas allows us to create timezone-aware datetime values.

## Example

We can create a timestamp using the Asia/Kolkata timezone.

```python
import pandas as pd

time = pd.Timestamp(
    "2026-10-08 10:30:00",
    tz="Asia/Kolkata"
)

print(time)
```

## Output Explanation

The timestamp contains the date, time, and Asia/Kolkata timezone.
Time zones are useful when working with international datasets.

# 6. Extracting Date Features

Date features are useful information extracted from datetime values.
Common features include year, month, day, weekday, and quarter.

## Example

From a date, we can extract the year, month, day, day name, and quarter.
These features can be useful for Machine Learning models.

```python
import pandas as pd

df = pd.DataFrame({
    "Date": pd.to_datetime([
        "2026-10-08",
        "2026-10-09",
        "2026-10-10"
    ])
})

df["Year"] = df["Date"].dt.year
df["Month"] = df["Date"].dt.month
df["Day"] = df["Date"].dt.day
df["Day_Name"] = df["Date"].dt.day_name()
df["Quarter"] = df["Date"].dt.quarter

print(df)
```

## Output Explanation

New columns are created from the Date column.
These features help identify time-based and seasonal patterns in data.
