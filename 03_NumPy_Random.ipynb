# Notebook 3 – NumPy Random Module

## 1. Random Numbers

Random numbers are values generated unpredictably within a specified range or distribution.

NumPy provides the `np.random` module to generate random values for simulations, testing, sampling, and Machine Learning tasks.

### Example

Suppose we want to simulate the results of rolling a dice multiple times.

Instead of manually entering the results, we can generate random values using NumPy.

In AI/ML, random numbers are also used for creating sample data and initializing models.

```python
import numpy as np

random_numbers = np.random.rand(5)

print("Random numbers:", random_numbers)
```

### Output Explanation

The output contains 5 random decimal values between 0 and 1.

The values will normally be different each time the code is executed.

## 2. Random Seed

A random seed is used to make random number generation reproducible.

When the same seed is used, NumPy generates the same sequence of random numbers.

### Example

Imagine two developers testing the same Machine Learning experiment.

Using the same random seed allows both developers to generate the same random values and reproduce the experiment.

```python
np.random.seed(42)

numbers = np.random.rand(5)

print("Random numbers:", numbers)
```

### Output Explanation

The seed `42` fixes the starting point of the random number generator.

Running the same code again with the same seed produces the same sequence of values.

## 3. Random Integers

The `np.random.randint()` function generates random integer values within a specified range.

The upper limit is excluded.

### Example

A dice has values from 1 to 6.

We can generate random dice results using random integers.

```python
np.random.seed(42)

dice = np.random.randint(1, 7, size=5)

print("Dice results:", dice)
```

### Output Explanation

The code generates 5 random integers between 1 and 6.

The value `7` is excluded because the upper limit of `randint()` is exclusive.

## 4. Random Choice

The `np.random.choice()` function randomly selects values from an existing array or sequence.

It can select one or multiple values.

### Example

Suppose a company has customers from different cities.

We can randomly select customers or cities for creating a sample dataset.

```python
cities = np.array(["Chennai", "Bangalore", "Mumbai", "Delhi"])

np.random.seed(42)

selected_city = np.random.choice(cities)

print("Selected city:", selected_city)
```

### Output Explanation

The code randomly selects one city from the given array.

Using a seed makes the selection reproducible.

## 5. Shuffle

The `np.random.shuffle()` function randomly changes the order of elements in an array.

It modifies the original array directly.

### Example

Suppose we have a list of students and want to randomly change their order before assigning them to groups.

Shuffling can help create a random arrangement.

```python
students = np.array(["Tony", "Arun", "John", "David", "Sam"])

np.random.seed(42)

np.random.shuffle(students)

print("Shuffled students:", students)
```

### Output Explanation

The elements are rearranged into a random order.

The original array is modified because `shuffle()` works directly on the array.

## 6. Permutation

The `np.random.permutation()` function returns a randomly rearranged copy of an array.

Unlike `shuffle()`, it does not modify the original array.

### Example

Suppose we have customer IDs and want to create a random order for analysis.

A permutation can create the randomized order while keeping the original data unchanged.

```python
customer_ids = np.array([101, 102, 103, 104, 105])

np.random.seed(42)

random_order = np.random.permutation(customer_ids)

print("Original:", customer_ids)
print("Permutation:", random_order)
```

### Output Explanation

The permutation creates a randomly reordered version of the array.

The original `customer_ids` array remains unchanged.

## 7. Random Distributions

A probability distribution describes how random values are spread or generated.

NumPy can generate random values from different distributions such as Normal, Uniform, Binomial, and Poisson distributions.

### Example

A company may simulate customer ages using a Normal distribution.

A random number between two fixed limits can be generated using a Uniform distribution.

Different distributions are useful for different real-world and AI/ML simulations.

```python
np.random.seed(42)

normal_data = np.random.normal(loc=50, scale=10, size=5)
uniform_data = np.random.uniform(low=10, high=20, size=5)

print("Normal distribution:")
print(normal_data)

print("\nUniform distribution:")
print(uniform_data)
```

### Output Explanation

The Normal distribution generates values around the specified mean of 50 with a standard deviation of 10.

The Uniform distribution generates values between 10 and 20 where values have equal probability across the range.
